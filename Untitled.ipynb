```python
# video
# якщо в параметрах вказати 0 - з'явиться картинка з веб-камери
video = cv2.VideoCapture("./assets/cat1.mp4")
```

```python
print(
    f"FPS: {video.get(cv2.CAP_PROP_FPS)}\n",
    f"width: {video.get(cv2.CAP_PROP_FRAME_WIDTH)}\n"
    f"height: {video.get(cv2.CAP_PROP_FRAME_HEIGHT)}\n"
    f"Frame count: {video.get(cv2.CAP_PROP_FRAME_COUNT)}\n"
)
```

```python
# Для камери

# video.set(cv2.CAP_PROP_FRAME_WIDTH, 400)
# video.set(cv2.CAP_PROP_FRAME_HEIGHT, 700)
```

```python
while True:
    success, frame = video.read()
    
    if not success: break

    frame_resize = cv2.resize(frame,(400,700))
    frame_resize = cv2.cvtColor(frame_resize, cv2.COLOR_BGR2GRAY)
    cv2.imshow("Video", frame_resize)
    if cv2.waitKey(60) & 0xFF in [ord('q'), ord('Q')]:
        print("Video stopped")
        break

video.release()
cv2.destroyAllWindows()
  # cv2_imshow(frame)
```
